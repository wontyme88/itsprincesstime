```python
#학교 안전 사고 데이터, 안전사고 보상 데이터 연도 별로 나누어서 읽기
import numpy as np
import pandas as pd

school_2021_df= pd.read_excel('★2021-2025 학교안전사고 데이터.xlsx', sheet_name='2021')
school_2022_df= pd.read_excel('★2021-2025 학교안전사고 데이터.xlsx', sheet_name='2022')
school_2023_df= pd.read_excel('★2021-2025 학교안전사고 데이터.xlsx', sheet_name='2023')
school_2024_df= pd.read_excel('★2021-2025 학교안전사고 데이터.xlsx', sheet_name='2024')
school_2025_df= pd.read_excel('★2021-2025 학교안전사고 데이터.xlsx', sheet_name='2025')

compe_2021_df= pd.read_excel('★2021-2025 학교안전사고 보상 데이터.xlsx',sheet_name='2021')
compe_2022_df= pd.read_excel('★2021-2025 학교안전사고 보상 데이터.xlsx',sheet_name='2022')
compe_2023_df= pd.read_excel('★2021-2025 학교안전사고 보상 데이터.xlsx',sheet_name='2023')
compe_2024_df= pd.read_excel('★2021-2025 학교안전사고 보상 데이터.xlsx',sheet_name='2024')
compe_2025_df= pd.read_excel('★2021-2025 학교안전사고 보상 데이터.xlsx',sheet_name='2025')
```

```python
#열(컬럼명)공백삭제
for df in [school_2021_df, school_2022_df, school_2023_df, school_2024_df, school_2025_df,compe_2021_df, compe_2022_df, compe_2023_df, compe_2024_df, compe_2025_df]: df.columns = df.columns.str.strip()
```

```python
#학교 안전 사고 데이터 자료 파악하기
school_dfs = {
    2021: school_2021_df,
    2022: school_2022_df,
    2023: school_2023_df,
    2024: school_2024_df,
    2025: school_2025_df
}

for year, df in school_dfs.items():
    print(f"\n===== school {year} =====")
    print("shape:", df.shape)
    print("columns:", df.columns.tolist())
    print("info:", df.info())
    print("describe:", df.describe())
    print("head:", df.head())
    print()

```

```python
#학교 안전 사고 보상 데이터 자료 파악하기
compe_dfs = {
    2021: compe_2021_df,
    2022: compe_2022_df,
    2023: compe_2023_df,
    2024: compe_2024_df,
    2025: compe_2025_df
}

for year, df in compe_dfs.items():
    print(f"\n===== school {year} =====")
    print("shape:", df.shape)
    print("columns:", df.columns.tolist())
    print("info:", df.info())
    print("describe:", df.describe())
    print("head:", df.head())
    print()

```

```python
school_2021_df["연도"]=2021
school_2022_df["연도"]=2022
school_2023_df["연도"]=2023
school_2024_df["연도"]=2024
school_2025_df["연도"]=2025

compe_2021_df["연도"]=2021
compe_2022_df["연도"]=2022
compe_2023_df["연도"]=2023
compe_2024_df["연도"]=2024
compe_2025_df["연도"]=2025
```

```python
#전체 데이터들 하나로 합치기

school_all_df = pd.concat([
    school_2021_df,
    school_2022_df,
    school_2023_df,
    school_2024_df,
    school_2025_df
], ignore_index=True)

compe_all_df = pd.concat([
    compe_2021_df,
    compe_2022_df,
    compe_2023_df,
    compe_2024_df,
    compe_2025_df
], ignore_index=True)


```

```python
#합친 데이터 보기
print("===== 사고 데이터 열 정보 확인하기 =====")
print(school_all_df.columns.tolist())

print("\n===== 보상 데이터 열 정보 확인하기 =====")
print(compe_all_df.columns.tolist())


```

```python
#합친 데이터 보기
print("===== 사고 데이터 열 정보 확인하기 =====")
print(school_all_df.info())

print("\n===== 보상 데이터 열 정보 확인하기 =====")
print(compe_all_df.info())
```

```python
print("===== 사고 데이터 결측치 개수 =====")
print(school_all_df.isnull().sum())

print("\n===== 사고 데이터 결측 비율(%) =====")
print((school_all_df.isnull().mean() * 100).sort_values(ascending=False))

print("\n===== 보상 데이터 결측치 개수 =====")
print(compe_all_df.isnull().sum())

print("\n===== 보상 데이터 결측 비율(%) =====")
print((compe_all_df.isnull().mean() * 100).sort_values(ascending=False))

```

```python
for col in school_all_df.columns:
    print(f"\n===== 사고 데이터: {col} =====")
    print(school_all_df[col].value_counts(dropna=False).head(20))
for col in compe_all_df.columns:
    print(f"\n===== 보상 데이터: {col} =====")
    print(compe_all_df[col].value_counts(dropna=False).head(20))


```

```python
school_cols = set(school_all_df.columns)
compe_cols = set(compe_all_df.columns)

common_cols = school_cols.intersection(compe_cols)

print("공통 컬럼:")
print(common_cols)
```
